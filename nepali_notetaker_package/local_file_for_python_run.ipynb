# Script to download the whisper and store it locally

```python
from transformers import WhisperFeatureExtractor, WhisperTokenizer, WhisperForConditionalGeneration
import os
```

```python
local_dir = r"D:\nepali_notetaker_project\nepali_notetaker_package\pretrained\nepali_whisper"

huggingface_model_id = "openai/whisper-small"

if not os.path.exists(local_dir):
    os.makedirs(local_dir)

print(f"Downloading {huggingface_model_id} to {local_dir}")

WhisperFeatureExtractor.from_pretrained(huggingface_model_id).save_pretrained(local_dir)
WhisperTokenizer.from_pretrained(huggingface_model_id).save_pretrained(local_dir)
WhisperForConditionalGeneration.from_pretrained(huggingface_model_id).save_pretrained(local_dir)

print(f"Model downloaded and saved to {local_dir}")

```

# Load the model 

```python
feature_extractor = WhisperFeatureExtractor.from_pretrained(local_dir)
tokenizer = WhisperTokenizer.from_pretrained(local_dir)
model = WhisperForConditionalGeneration.from_pretrained(local_dir)

# Test the model
audio_path = "D:\nepali_notetaker_project\nepali_notetaker_package\DataSet\asr_nepali_0\asr_nepali\data\00\processed_wav\000a0be2c6.wav"



```
